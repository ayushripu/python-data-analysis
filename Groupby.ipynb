```python
import pandas as pd

df = pd.DataFrame({
    'Name' : ['Ayush', 'Ved', 'Ravi', 'Hemant', 'Aryan', 'Ayush', 'Ravi'],
    'Age'  : [24,45,29,30,55, 36, 22],
    'Salary':[30000,40000,20000,19500,13500, 6734, 3467]
})
df
```

```python
df1 = df.groupby('Name')
df1
```

```python
for x,y in df1:
    
    print(x)
    print(y)
    
```

```python
df2 = df.groupby('Name')['Salary'].max()
df2
```

```python
df3 = df.groupby('Name')['Salary'].min()
df3 
```

```python
df4 = df.groupby('Name')['Salary'].mean()
df4
```

```python
df5 = df.groupby('Name')['Salary'].count()
df5
```

```python
df6 = df.groupby('Name').count()
df6
```

```python
df7 = df.groupby('Name')['Salary'].sum()
df7
```

```python
df8 = df.groupby('Name')['Salary'].agg(['sum', 'max', 'min', 'mean', 'count'])
df8
```

```python
df9 = df.groupby('Name')['Salary'].agg(['max'])
df9
```
