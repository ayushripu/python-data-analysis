# WAP to solve the given below Requeriment

                input: a4b6c7
                output: aebhcj

```python
st = "a4b6c7"
out = ""

for i in st:
    if i.isalpha():
        out = out+i
        ch = i
    else:
        d = int(i)
        out = out+chr(ord(ch)+d)

print(out)
```

# WAP to solve the given below Requeriment
    Input:  aaaabbbzz
    Output: a4b3z2

```python
s = "aaaabbbzz"
# s = input("Enter a string: ")

count = 1

for i in range(len(s) - 1):
    if s[i] == s[i + 1]:
        count += 1
    else:
        print(s[i], count, sep="", end="")
        count = 1

# Print the last character and its count
print(s[-1], count, sep="")
```

# `Transfer Statement`

:-> Transfer statement takes the controller from one point of the program to another point of the program

           . There are 3 Types of Transfer Statement

                1) Break
                2) Continue
                3) Pass
1) `Break:-> ` Break is a keyword which terminate the loop as soon as it encounter it.

                                       OR

               Break is a keyword which takes the controller out of current working loop.

                                       

   

3) `Continue:->`  Continue is a Keyword which is used to skip the current iteration and controller jump to the next iteration

4) `Pass:->` Pass is a keyword with help us to create empty block in python                                   

```python
#Break

i = 1
while i <= 5:
    if i ==  3:
        break
    print(i)
    i+=1
```

```python
#Nested loop break

i = 1
while i <= 5:
    j = 1
    while j <= 5:
        if j == 3:
            break
        print(i,",",j)
        j+=1
    i+=1
```

```python
# Continue

for i in range(1,6):
    if i == 3:
        continue
    print(i)
```

```python
# Nested Loop for Continue
for i in range(1,6):
    for j in range(1,6):
        if j == 3:
            continue
        print(i,",",j)
```

```python
i = 1
while i <= 10:
    pass
i+=1
```
