# Filling Null Values

**fillna()**

```python
df1 = df.fillna(value = 0)
df1
```

```python
df1.isnull().sum().sum()
```

```python
df2 = df.fillna(value = 5)
df2
```

**Filling Null values with a previous value**

```python
df3 = df.fillna(method = 'pad')
df3
```

**Filling Null values with a next value**

```python
df4 = df.fillna(method = 'bfill')
df4
```

```python
df5 = df.fillna(method = 'pad', axis = 1)
df5
```

```python
df6 = df.fillna(method = 'bfill', axis = 1)
df6
```

```python
df
```

**Filling Different values in null in different columns**

```python
df7 = df.fillna({'society' : 'asdf', 'balcony' : 'jkl;'})
df7
```

**Filling null value with the mean of a columns**

```python
df8 = df.fillna(value=df['balcony'].mean())
df8
```

# DROP Null Values

**dropna()**

```python
df9 = df.dropna()
df9
```

```python
df10 = df.dropna(how= 'any')
df10
```

```python
df11 = df.dropna(how= 'all')
df11
```

## Replace Value

**replace()**

```python
import numpy as np

df12 = df.replace(to_replace=np.nan, value= 8978)
df12
```

```python

df13 = df.replace(to_replace=3.0, value= 5.0)
df13
```
